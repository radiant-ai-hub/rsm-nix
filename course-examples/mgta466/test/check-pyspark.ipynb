# Check PySpark

Run this in `mgta466/examples/test` with the **Python (Spark/Hadoop)** kernel. The local session uses one CPU worker so it is suitable for the shared MSBA server.


```python
# test pyspark
import pyspark
from pyspark.sql import SparkSession
```

```python
spark = (SparkSession.builder.master("local[1]")
         .appName("test-pyspark")
         .config("spark.ui.enabled", "false")
         .config("spark.driver.memory", "1g")
         .getOrCreate())
data = [("John", 28), ("Smith", 35), ("Sarah", 29)]
columns = ["Name", "Age"]
df = spark.createDataFrame(data, columns)
df.show()
```

```python
spark.stop()
```

```python
import pyspark
from pyspark.sql import SparkSession

conf = pyspark.SparkConf().setAll(
    [("spark.master", "local[1]"), ("spark.app.name", "Basic Setup")]
)
spark = SparkSession.builder.config(conf=conf).getOrCreate()

print(pyspark.__version__)
print(spark.version)
spark.stop()
```
