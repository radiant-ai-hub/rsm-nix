```python
import sys
import numpy as np
import polars as pl

print('python :', sys.executable)
print('numpy  :', np.__version__)
print('polars :', pl.__version__)
```

## A Polars DataFrame

```python
df = pl.DataFrame({'x': range(1, 11), 'y': [i ** 2 for i in range(1, 11)]})
df
```

## An inline plot (plotnine)

```python
from plotnine import ggplot, aes, geom_line, geom_point, labs

(
    ggplot(df, aes('x', 'y'))
    + geom_line()
    + geom_point()
    + labs(title='y = x^2', x='x', y='y')
)
```
