# pyrsm in a notebook

A linear regression with **pyrsm** (the Rady MSBA helper package) on synthetic
Polars data, using the R-style formula interface. Run all cells.

```python
import numpy as np
import polars as pl
import pyrsm as rsm

rng = np.random.default_rng(42)
n = 300
x1 = rng.normal(size=n)
x2 = rng.normal(size=n)
price = 5 + 2.0 * x1 - 1.0 * x2 + rng.normal(scale=0.5, size=n)
df = pl.DataFrame({'x1': x1, 'x2': x2, 'price': price})
df.head()
```

## Regression (formula interface)

```python
reg = rsm.model.regress(data={'df': df}, formula='price ~ x1 + x2')
reg.summary()
```

## Visualize the relationship

```python
from plotnine import ggplot, aes, geom_point, geom_smooth, labs

(
    ggplot(df, aes('x1', 'price'))
    + geom_point(alpha=0.4)
    + geom_smooth(method='lm')
    + labs(title='price vs x1', x='x1', y='price')
)
```
